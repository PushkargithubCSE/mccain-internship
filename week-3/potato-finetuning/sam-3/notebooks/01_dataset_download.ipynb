```python
import os
from dotenv import load_dotenv
from roboflow import Roboflow

env_path = os.path.join(PROJECT_PATH, ".env")
load_dotenv(env_path)
api_key = os.getenv("ROBOFLOW_API_KEY")
if not api_key:
    raise RuntimeError("ROBOFLOW_API_KEY is missing from the project .env file")

print("Key loaded:", bool(api_key))

rf = Roboflow(api_key=api_key)
project = rf.workspace("pushkar-chandra").project("segmentation-jajrd")
dataset = project.version(2).download("coco-segmentation")

print(f"Dataset ready at: {dataset.location}")
```

```python
import json

for split in ['train', 'valid', 'test']:
    ann_path = f"{dataset.location}/{split}/_annotations.coco.json"
    if os.path.exists(ann_path):
        with open(ann_path) as f:
            coco_data = json.load(f)
        
        num_with_segmentation = sum(1 for ann in coco_data['annotations'] if len(ann.get('segmentation', [])) > 0)
        
        print(f"[{split.upper()}]")
        print(f"  Images: {len(coco_data['images'])}")
        print(f"  Annotations: {len(coco_data['annotations'])}")
        print(f"  Annotations WITH polygon segmentation: {num_with_segmentation}")
        print()
```
