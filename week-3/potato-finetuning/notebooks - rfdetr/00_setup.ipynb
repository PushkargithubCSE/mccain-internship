```python
import os

PROJECT_PATH = "/content/drive/MyDrive/mccain-internship/week-3/potato-finetuning"

# Create folder structure
os.makedirs(f'{PROJECT_PATH}/src/utils', exist_ok=True)

colab_utils_code = '''"""Colab setup utilities."""
import os, sys, torch

DRIVE_PROJECT_PATH = "/content/drive/MyDrive/mccain-internship/week-3/potato-finetuning"

def setup_colab(project_path: str = DRIVE_PROJECT_PATH):
    src_path = os.path.join(project_path, "src")
    if src_path not in sys.path:
        sys.path.append(src_path)
    print(f"✓ Project root: {project_path}")
    return project_path

def ensure_folders(project_path: str = DRIVE_PROJECT_PATH):
    folders = [
        "data/raw", "data/processed", "data/splits",
        "src/data", "src/models", "src/training",
        "src/evaluation", "src/pipeline", "src/utils",
        "notebooks", "configs",
        "checkpoints/linear_probe", "checkpoints/full_finetune", "checkpoints/lora",
        "experiments", "report", "tests",
    ]
    for folder in folders:
        os.makedirs(os.path.join(project_path, folder), exist_ok=True)
    print(f"✓ Folder structure created")

def gpu_check():
    if torch.cuda.is_available():
        print(f"✓ GPU: {torch.cuda.get_device_name(0)}")
        print(f"  Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
    else:
        print("✗ No GPU available")
'''

with open(f'{PROJECT_PATH}/src/utils/colab_utils.py', 'w') as f:
    f.write(colab_utils_code)

print("✓ colab_utils.py created")
```
